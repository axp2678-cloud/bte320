```python

#B
def loan_payment(principal, annual_rate, months):
    assert (isinstance(principal, (int, float)) and
            isinstance(annual_rate, (int, float)) and
            isinstance(months, int))
    if (principal <= 0) or (annual_rate <= 0) or (months <= 0):
        raise ValueError("Invalid Input")
    try:
        r = annual_rate / 12
        if r == 0:
            pmt = principal / months
        else:
            pmt = r * principal / (1 - (1 + r) ** (-months))
        return pmt
    except Exception as e:
        print(f"Error: {e}")
```

quality control manager

```python

def getValues(n):
    values = []
    for i in range(n):
        while True:
            try:
                val = float(input(f"Enter value {i+1}: "))
                if val < 0:
                    raise ValueError("Negative value not allowed")
                values.append(val)
                break
            except ValueError as e:
                print(f"Invalid input: {e}")
    return values

def calcAverage(alist):
    if len(alist) <= 3:
        raise ValueError("Not enough values to exclude first three")
    trimmed = alist[3:]
    return sum(trimmed) / len(trimmed)

n = int(input("How many values do you want to enter? "))
values = getValues(n)
print(f"Average (excluding first 3): {calcAverage(values)}")
```

greeting class

```python

class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(f"{self.first}")

    def __str__(self):
        return f"{self.first} {self.last}"

g = Greeting("Audrey", "Purnell")
g.display(True)
g.display(False)
print(g)
```

student class

```python
class Student:
    def __init__(self, name, number):
        self.__name = name
        self.number = number
        self.__courses = []

    def get_name(self):
        return self.__name

    def get_courses(self):
        return self.__courses

    def set_name(self, newName):
        self.__name = newName

    def __len__(self):
        return len(self.__courses)

    def enroll(self, newCourse):
        if newCourse not in self.__courses:
            self.__courses.append(newCourse)
            print(f"Enrolled in {newCourse}")
        else:
            print(f'Student {self.__name} has already enrolled in {newCourse}')
```

classes point and rectangle

```python

import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)

class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4
        self.width = self.calcWidth()
        self.height = self.calcHeight()

    def calcWidth(self):
        return self.p1.distance(self.p2)

    def calcHeight(self):
        return self.p2.distance(self.p3)

    def calcArea(self):
        return self.width * self.height


p1 = Point(0, 0)
p2 = Point(4, 0)
p3 = Point(4, 3)
p4 = Point(0, 3)

rect = Rectangle(p1, p2, p3, p4)
print(f"Width: {rect.calcWidth()}")
print(f"Height: {rect.calcHeight()}")
print(f"Area: {rect.calcArea()}")
```

employee manager class

```python

class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__wage = wage
        self.hours = hours

    def get_wage(self):
        return self.__wage

    def set_wage(self, new_wage):
        if new_wage < 0:
            print("Error: negative wages not allowed")
        else:
            self.__wage = new_wage

    def calculate_pay(self):
        return self.__wage * self.hours

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Hourly Wage: ${self.get_wage():.2f}")
        print(f"Hours Worked: {self.hours}")
        print(f"Weekly Pay: ${self.calculate_pay():.2f}")


class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):
        return super().calculate_pay() + self.bonus

    def display_info(self):
        super().display_info()
        print(f"Bonus: ${self.bonus:.2f}")

print("=== Employee ===")
emp = Employee("John Smith", 15.00, 40)
emp.display_info()

print("\n=== Manager ===")
mgr = Manager("Jane Doe", 25.00, 40, 500)
mgr.display_info()
```
